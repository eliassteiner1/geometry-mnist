# Generating a chain from braid-closure words

### individual chain segments

```python
import numpy as np
import einops as eo

from geometry_mnist.knotgen.braidclosure import _generate_braid_blocks
from geometry_mnist.plotly.plots import plot_bpts


word = [1,-2,1,3,2,2,-4,-3,2,-3,-3,-4] # word must be 1-indexed to hold sign+braid_id. 
n_pts_per_ops = 50
n_braids = np.max(np.abs(word)) + 1
n_ops = len(word)

bpts = _generate_braid_blocks(word, n_braids, n_ops, n_pts_per_ops)
bpts = eo.rearrange(bpts, "braid (block pts) dim -> (braid block) pts dim", pts=n_pts_per_ops)
plot_bpts(bpts)
```

### the full raw braid as one contiguous chain

```python
import numpy as np

from geometry_mnist.knotgen.braidclosure import generate_braid_closure
from geometry_mnist.plotly.plots import plot_pts


word = [1,-2,1,3,2,2,-4,-3,2,-3,-3,-4] # word must be 1-indexed to hold sign+braid_id. 
n_pts_per_ops = 30

pts_raw = generate_braid_closure(word, n_pts_per_ops)
plot_pts(pts_raw)
```

### resampled and normalized chain

```python
from geometry_mnist.knotgen.braidclosure import generate_braid_closure
from geometry_mnist.knotgen.braidclosure import resample_normalize_chain
from geometry_mnist.plotly.plots import plot_pts


word = [1,-2,1,3,2,2,-4,-3,2,-3,-3,-4] # word must be 1-indexed to hold sign+braid_id. 
n_pts_per_ops = 30

pts_raw       = generate_braid_closure(word, n_pts_per_ops)
pts_resampled = resample_normalize_chain(pts_raw, n_resample=512)

plot_pts(pts_resampled, kos_len=0.05)
```
